```python
print("\n========== GROUP-BY ANALYSIS ==========")

# Automatically find important categorical columns
# and numerical columns

for cat_column in categorical_columns:

    print("\n====================================")
    print("Analysis by:", cat_column)
    print("====================================")

    # Mean of all numerical variables
    grouped_mean = df.groupby(cat_column)[numerical_columns].mean()

    display(grouped_mean)
```

Correlation matrix

```python
print("\n========== CORRELATION MATRIX ==========")

correlation_matrix = df[numerical_columns].corr()

display(correlation_matrix)


# ============================================
# 21. CORRELATION HEATMAP
# ============================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python

print("\n========== STRONG CORRELATIONS ==========")

# Convert correlation matrix into pairs
corr_pairs = correlation_matrix.unstack()

# Remove self-correlations
corr_pairs = corr_pairs[corr_pairs.index.get_level_values(0)
                        != corr_pairs.index.get_level_values(1)]

# Remove duplicate pairs
corr_pairs = corr_pairs.sort_values(ascending=False)

print(corr_pairs.head(10))
```

   each observation is unique.

3. Customer activity varies across the dataset, with differences

4. Revenue and profit vary considerably across branches, indicating
   differences in overall branch performance.

5. Customer ratings and delivery times provide useful information
   about customer experience and operational efficiency.

6. Marketing expenditure can be compared with revenue and profit
   to understand whether higher marketing investment is associated
   with better financial performance.

7. The correlation matrix shows the strength and direction of
   relationships between numerical variables. Variables with
   correlation values close to +1 or -1 have stronger relationships,
   while values close to 0 indicate weaker relationships.

8. Branch, region and store type comparisons help identify which
